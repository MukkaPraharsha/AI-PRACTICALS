```python
map_rooms = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": []
}

print("Our map:")
print("        A")
print("       / \\")
print("      B   C")
print("       \\ /")
print("        D")
print("        |")
print("        G (Treasure 💎)")

print("Doors from each room:", map_rooms)
```

```python
def bfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop(0)

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order


bfs_order = bfs("A", "G")

print("BFS checked the rooms in the order:", bfs_order)
print("Number of rooms BFS checked:", len(bfs_order))
```

```python
def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        if room == goal:
            return order

        for nxt in map_rooms[room]:
            to_do.append(nxt)

    return order


dfs_order = dfs("A", "G")

print("DFS checked the rooms in the order:", dfs_order)
print("Number of rooms DFS checked:", len(dfs_order))
```

```python
door_cost = {
    ("A", "B"): 1,
    ("B", "D"): 1,
    ("A", "C"): 5,
    ("C", "D"): 1,
    ("D", "G"): 1
}


def path_cost(path):
    total = 0

    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total += door_cost[door]

    return total


path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

print("Path 1 A -> B -> D -> G costs:",
      path_cost(path1))

print("Path 2 A -> C -> D -> G costs:",
      path_cost(path2))
```

```python
cost1 = path_cost(path1)
cost2 = path_cost(path2)

# We want the cheaper path
if cost1 < cost2:
    best_path = path1
    best_cost = cost1
else:
    best_path = path2
    best_cost = cost2

print("Path 1 cost:", cost1)
print("Path 2 cost:", cost2)
print("The cheapest path is:", best_path,
      "with cost", best_cost, "💎")

bfs_order = bfs("A", "G")
dfs_order = dfs("A", "G")

print("RESULTS")
print("-" * 40)

print("BFS (nearest first) checked:",
      len(bfs_order), "rooms ->", bfs_order)

print("DFS (deep first) checked:",
      len(dfs_order), "rooms ->", dfs_order)

print("Cheapest path:", best_path,
      "with cost", best_cost)

print("-" * 40)

print("What we learned:")
print("* BFS checks the nearest rooms first (spreads out).")
print("* DFS dives deep down one path first.")
print("* Cheapest path counts door costs and picks the lowest total.")
```

```python
door_cost[("A", "B")] = 8
door_cost[("A", "C")] = 2
door_cost[("B", "D")] = 1
door_cost[("C", "D")] = 5
door_cost[("D", "G")] = 1

path1_cost = (
    door_cost[("A", "B")]
    + door_cost[("B", "D")]
    + door_cost[("D", "G")]
)

path2_cost = (
    door_cost[("A", "C")]
    + door_cost[("C", "D")]
    + door_cost[("D", "G")]
)

print("Now path 1 (through B) costs:", path1_cost)
print("Now path 2 (through C) costs:", path2_cost)

if path1_cost < path2_cost:
    print("Cheapest path is now Path 1: A -> B -> D -> G")
else:
    print("Cheapest path is now Path 2: A -> C -> D -> G")
```
